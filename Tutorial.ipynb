# Template Matching - OpenCV

```python
import cv2
import numpy as np
```

```python
scr_img = cv2.imread('./images/game.png', cv2.IMREAD_UNCHANGED)
wood_img = cv2.imread('./images/woodleft.png', cv2.IMREAD_UNCHANGED)
scr_img = scr_img[:, :, :3]
wood_img = wood_img[:, :, :3]
```

```python
cv2.imshow('Farm', scr_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imshow('Needle', wood_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Template Matching
<ul>
    <li>There are 6 comparison methods</li>
    <ul>
        <li>TM_CCOEFF</li>
        <li>TM_CCOEFF_NORMED</li>
        <li>TM_CCORR</li>
        <li>TM_CCORR_NORMED</li>
        <li>TM_SQDIFF</li>
        <li>TM_SQDIFF_NORMED</li>
    </ul>
    <li><a href="https://docs.opencv.org/master/d4/dc6/tutorial_py_template_matching.html">OpenCV Template Matching Docs</a></li>
</ul>

```python
result = cv2.matchTemplate(scr_img, wood_img, cv2.TM_CCOEFF_NORMED)
result
```

```python
cv2.imshow('Result', result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Get Max Result

```python
min_val, max_val, min_loc, max_loc = cv2.minMaxLoc(result)
max_loc
```

```python
max_val
```

### Paint Square on Main Image

```python
w = wood_img.shape[1]
h = wood_img.shape[0]
```

```python
cv2.rectangle(scr_img, max_loc, 
              (max_loc[0] + w, max_loc[1] + h), (0, 255, 255), 2)
```

### Paint Multiple Squares

```python
threshold = .85  # 阈值
yloc, xloc = np.where(result >= threshold)
len(xloc)
```

```python
for (x, y) in zip(xloc, yloc):
    cv2.rectangle(scr_img, (x, y), (x + w, y + h), (0, 255, 255), 2)
```

```python
cv2.imshow('Farm', scr_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
rectangles = []
for (x, y) in zip(xloc, yloc):
    rectangles.append([int(x), int(y), int(w), int(h)])
    rectangles.append([int(x), int(y), int(w), int(h)])
len(rectangles)
```

```python
rectangles, weights = cv2.groupRectangles(rectangles, 1, 0.2)
rectangles
```
