```python
import cv2
```

```python
img = cv2.imread('./images/landscape.jpg')
img
```

```python
print(type(img))
print(img.shape)
```

```python
cv2.imshow('Original', img)
cv2.waitKey(0)
```

```python
small = cv2.resize(img, (0, 0), fx=0.2, fy=0.2)
cv2.imshow('Small', small)
cv2.waitKey(0)
```

```python
small.shape
```

```python
small[0][0]
small[0:100][0:100] = (255, 255, 255)
small[0][0]
```

```python
cv2.imshow('Small', small)
cv2.waitKey(0)
```

```python
gray = cv2.cvtColor(small, cv2.COLOR_BGR2GRAY)
gray.shape
```

```python
cv2.imshow('gray', gray)
cv2.waitKey(0)
```

```python
cv2.imwrite('./images/Gray.png', gray)
```
